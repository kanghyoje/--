## 타이타닉 기계학습

```python
import pandas as pd
titanic = pd.read_csv('train.csv')
print(titanic.isnull().sum())
```

```python
avg = titanic['Age'].mean()
titanic['Age'] = titanic['Age'].fillna(avg)
print(titanic.isnull().sum())
```

```python
titanic.drop(['Cabin'], axis = 1, inplace = True)
print(titanic.isnull().sum())
```
